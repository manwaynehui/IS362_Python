## Question 2 - What is the easternmost airport in the United States?

* Sort Values: Sort by `lon` with `ascending=False` to bring the highest longitudes to the top.
* Select Columns: `faa`, `name`, `lat`, and `lon` for the top 5 results.

First, I looked up Eareckson As airport from the cordinates in the dataframe and it showed me the most west part of the United States. Then I saw airports like MYF (San Diego), DVT (Phoenix), and EEN at the top, but a quick search revealed they are data errors missing negative signs. After couple quick search of the 5 airports there are two technical answers.

**Answer:**

* `By Geographic Coordinates: Eareckson As (SYA)` on Shemya Island, is technically the easternmost U.S. airport because it crosses the 180° mark, its longitude flips to **+174.11° E**, placing it inside the Eastern Hemisphere.
* `By Conventional Geography (U.S. Map): Eastport Municipal Airport (EPM)` in Maine is the easternmost airport at **-67.01° W**, representing the furthest eastern point of the United States from looking at the US center point.

```python
feb12_weather = weather[
    (weather["year"] == 2013) & 
    (weather["month"] == 2) & 
    (weather["day"] == 12)
]


feb12_weather.sort_values(by="wind_speed", ascending=False).head(5)[
    ["origin", "time_hour", "wind_speed"]
]
```

```python
clean_feb12 = feb12_weather[feb12_weather["wind_speed"] < 100]

wind_summary = clean_feb12.groupby("origin").agg(
    avg_wind_speed=("wind_speed", "mean"),
    max_wind_speed=("wind_speed", "max")
).reset_index()

wind_summary
```

## Question 3 - On February 12th, 2013, which New York area airport had the windiest weather?

* Filter the `weather` dataset for February 12, 2013, and sort by `wind_speed` descending to view the top recorded wind speeds.

* At first glance, **EWR** records an impossible wind speed of **1,048.36 mph**, which is an obvious weather sensor glitch.

* Filter out readings over 100 mph (`wind_speed < 100`) to remove the corrupted data, then group by `origin` to calculate the daily average and peak wind speeds across the three airports.

* `.agg()` function compute multiple summary statistics (such as the average and maximum) across grouped columns all at once in a single clean table.

* `.groupby("origin")` splits the 24 hourly weather observations into separate groups for each airport (EWR, JFK, and LGA) so we can calculate summary statistics for it individually.

**Answer:**
After dropping the sensor error at EWR, **LaGuardia Airport (LGA)** had the windiest weather on February 12, 2013, with the highest average wind speed (~14.96 mph) and the highest sustained peak wind speed (~23.02 mph).
