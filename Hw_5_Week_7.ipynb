## 1. Loading Rating Data
* First we import the `pandas` library and load our movies rating data we collected from the CSV file.
* To make row calculation easier we set the `User` column as the index of the DataFrame.
* Lastly we displayed the data to verify it loaded correctly.

```python
import pandas as pd

df = pd.read_csv('movie_ratings.csv') 
df = df.set_index('User')

display(df)
```

## 2. Cacluate Average Ratings 
* We need to calculate the averages per movie and per user by using the columns and rows.
* `df.mean()` calculates the average down the columns and ignores the `NaN` values.
* `df.mean(axis=1)` calculates the average across rows for each individual users.

```python
print("\n ---- Average Rating Per Movie (Column Averages) ----\n")
movie_averages = df.mean()
print(movie_averages)

print("\n ---- Average Rating Per User (Row Averages) ----\n")
user_averages = df.mean(axis=1)
print(user_averages)
```

## Normalize Ratings
To account for individual user bias, we can use the **Min-Max Normalization** row-by-row. 

We use `lambda` function to apply the formula from the article (https://bi-analytics.org/forums/topic/9-standardization-vs-normalization/). 

`X = ([x - x.min] / [x.max - x.min])`

*   `lambda x`: Creates a temporary function where `x` represents a single row (one user's ratings).
*   `x.min()` and `x.max()`: Finds the lowest and highest score that specific user gave.
*   `(x - x.min())`: Subtracts their lowest score from all their ratings. This shifts their data so their lowest score becomes `0`.
*   `/ (x.max() - x.min())`: Divides by the range (the difference between their highest and lowest score). This scales their data so their highest score becomes `1`.
*   `if x.max() != x.min() else x`: This prevents a division-by-zero error just in case a user accidentally gave the exact same score to every movie.
*   `axis=1`: Tells `df.apply()` to run this function across the rows (users) rather than the columns.


## Pros and Cons of Normalized Ratings

**Advantages of Normalized Ratings:**
*   **Removes User Bias:** Normalizing reveals a user's *relative* preferences. It shows which movies they liked best compared to the others they saw, rather than just looking at raw scores, making it easier to compare a "harsh grader" with an "easy grader."
*   **Better Comparisons:** It allows a recommender system to compare users fairly, regardless of how they individually use the 1-5 scale.

**Disadvantages of Normalized Ratings:**
*   **Loses Absolute Meaning:** A `1.0` in a normalized dataframe just means it was that user's *favorite* movie from the list. However, it doesn't tell us if they actually thought the movie was objectively good as they might have rated their favorite a 3 out of 5.
*   **Harder to Interpret:** Raw scores like a 5 out of 5 are intuitive to humans. Normalized scores like `0.875` require knowing the original minimum and maximum to understand what they represent.


```python
df_normalized = df.apply(lambda x: (x - x.min()) / (x.max() - x.min()) if x.max() != x.min() else x, axis=1)

print("\n---- Normalized Ratings (Per User) ----\n")
display(df_normalized)

print("\n---- Average of Normalized Ratings Per Movie ----\n")
print(df_normalized.mean())
```

## Standardize Ratings 
Similar to normalization, we can use **Standardization (Z-score)**. 

We use another `lambda` function to apply the formula from the article (https://bi-analytics.org/forums/topic/9-standardization-vs-normalization/): 

`(x - mean) / standard_deviation`

*   `lambda x`: Creates a temporary function representing a single row (one user's ratings).
*   `x.mean()`: Calculates the average score that specific user gave across all movies they rated.
*   `(x - x.mean())`: Subtracts their average from every rating. This *centers* their data. If they gave a movie exactly their average score, it becomes `0`. If they gave it a higher score, it becomes a positive number; if lower, a negative number.
*   `x.std()`: Calculates the standard deviation (how spread out their ratings are).
*   `/ x.std()`: Divides by the standard deviation. This scales the data so that the numbers represent "how many standard deviations away from their average" a rating is.
*   `axis=1`: Tells `df.apply()` to run this row-by-row.

```python
df_standardized = df.apply(lambda x: (x - x.mean()) / x.std(), axis=1)

print("\n---- Standardized Ratings (Per User) ---- \n")
display(df_standardized)

print("\n---- Average of Standardized Ratings Per Movie ----\n")
print(df_standardized.mean())
```
