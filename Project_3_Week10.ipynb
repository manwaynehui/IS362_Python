# Project 3: Loading Relational SQL Data into pandas

## Objective:
- Connect Relationsal Database Maanagement System.
- Inspect the relational schema.
- Query mutli-table data using pandas dataFrame using an SQL to join five tables.

```python
import pandas as pd
import sqlalchemy as sqla


db = sqla.create_engine('sqlite:///Chinook_Sqlite.sqlite')
```

## Database Connection
To be able to manipulate Chinook dataBase, we import `sqlAlchemy` and use `create_engine()` function:
- The `create_engine()` act as the central connection interface that routes between our `pandas` and `SQLite`.

```python
inspector = sqla.inspect(db)

for table in inspector.get_table_names():
    display(pd.read_sql(f"SELECT * FROM {table} LIMIT 0;", db).iloc[:,:1])
```

### DataBase Scehme & Relationship Discovery
Before making any queries, we musst analyze our tables and headers to get a understanding of the relationships.
- `sqla.inspect(db)` get  schema.
- `for table in inspector.get_table_names():` loop so we can gather data each table
    - `display(pd.read_sql(f"SELECT * FROM {table} LIMIT 0;", db).iloc[:0,:1])` limits the row to 0, 
        - use `.iloc[:,:1]` to get just the header and leaves the row alone.

```python
sql_table = """
SELECT 
    Customer.LastName,
    Customer.FirstName,
    Track.Name,
    Album.Title
FROM Customer
JOIN Invoice     USING (CustomerId)
JOIN InvoiceLine USING (InvoiceId)
JOIN Track       USING (TrackId)
JOIN Album       USING (AlbumId)
ORDER BY 
    Customer.LastName, 
    Customer.FirstName;

""" 


df = pd.read_sql(sql_table, db).head(5)
df
```
