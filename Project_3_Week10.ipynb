```python
import pandas as pd
import sqlalchemy as sqla


db = sqla.create_engine('sqlite:///Chinook_Sqlite.sqlite')
```

## Database Connection
To be able to manipulate Chinook dataBase, we import `sqlAlchemy` and use `create_engine()` function:
- The `create_engine()` act as the central connection interface that routes between our `pandas` and `SQLite`.

```python
inspector = sqla.inspect(db)

for table in inspector.get_table_names():
   display(pd.read_sql(f"SELECT * FROM {table} LIMIT 0;", db))

```

### DataBase Schema & Relationship Discovery
Before making any queries, we musst analyze our tables and headers of each column to get a understanding of the relationships. Being able to see the headers allows us to match how the tables relate to each other by their shared column names.

- `sqla.inspect(db)` Connects an inspector tool to check the database layout to read    the table names and columns.
- `for table in inspector.get_table_names():` loops through each table in the database
    - `display(pd.read_sql(f"SELECT * FROM {table} LIMIT 0;", db))` 
        - `SELECT * FROM {table} LIMIT 0` request information from database and limits the row of table to 0, which only shows the header.



```python
sql_table = """
SELECT 
    c.LastName,
    c.FirstName,
    t.Name,
    a.Title
FROM Customer AS c
JOIN Invoice     AS i  ON c.CustomerId  = i.CustomerId
JOIN InvoiceLine AS il ON i.InvoiceId   = il.InvoiceId
JOIN Track       AS t  ON il.TrackId    = t.TrackId
JOIN Album       AS a  ON t.AlbumId     = a.AlbumId
ORDER BY 
    c.LastName, 
    c.FirstName;

""" 


df = pd.read_sql(sql_table, db).head(5)
df
```

## Query Database

To display each customer's purchased tracks alongside their respective album titles, we query across five related tables by following the database relationships. Now that we know from our previous inspection code which columns have matching names, we can use SQL to join the information together and output the requested table (Last Name, First Name, Track Name, Album Title).

`SELECT:` Extracts the following data:
- Customer's LastName & FirstName
- The track name
- Album title

`FROM Customer AS c:` 
- Starts at the Customer table (c) to capture customer names.

`JOIN Invoice AS i ON c.CustomerId = i.CustomerId:` 
- Finds the matching "invoice" for each "customer" using CustomerId.

`JOIN InvoiceLine AS il ON i.InvoiceId = il.InvoiceId:` 
- Finds the matching "purchased items" for each "invoice" using InvoiceId.

`JOIN Track AS t ON il.TrackId = t.TrackId:` 
- Finds the matching "song" for each "purchased item" using TrackId.

`JOIN Album AS a ON t.AlbumId = a.AlbumId:` 
- Finds the matching "album" for each "song" using AlbumId.

`ORDER BY c.LastName, c.FirstName:`
- Sorts the output alphabetically by customer last name, then by first name.

`df = pd.read_sql(sql_table, db).head(5)`
- Queries the SQL database and loads the entire returned table into a pandas DataFrame.
- keeps only first 5 row.
