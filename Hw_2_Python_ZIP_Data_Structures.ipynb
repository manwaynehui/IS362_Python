# Data Structure Topic: The zip() Function

## What is zip()?
The `zip()` function takes two or more lists and pairs up their matching items index-by-index, like a zipper on a jacket.

```python
# Creating two lists
students = ["Alice", "Bob", "Charlie", "Diana"]
scores = [85, 92, 78, 90]

# Combining them with zip()
paired_data = list(zip(students, scores))
print("Paired Data:", paired_data)
```

## Using zip() in a Loop
The most common way to use `zip()` is inside a loop to print or handle paired data together.

```python
# Looping through both lists at once
for student, score in zip(students, scores):
    print(f"{student} scored {score} points.")
```

## Unzipping Data with `*`
We can reverse the `zip()` process by passing zipped tuples into `zip()` with the `*` (unpacking) operator. This splits paired data back into separate tuples.

```python
# A list of paired tuples
pairs = [('Alice', 85), ('Bob', 92), ('Charlie', 78)]

# Unzip into two separate tuples using *
names, scores = zip(*pairs)

print("Unzipped Names:", names)
print("Unzipped Scores:", scores)
```

## Mismatched Lengths
When zipping sequences of unequal length, `zip()` stops processing as soon as the shortest sequence ends. Extra elements in longer sequences are ignored without throwing an error.

```python
# Lists of unequal lengths
students = ["Alice", "Bob", "Charlie", "Diana"]
scores = [85, 92]  # Only two scores!

# zip() stops at the shortest list, ignoring Charlie and Diana
truncated_pairs = list(zip(students, scores))
print("Truncated Result:", truncated_pairs)
```

## Preventing Data Loss with `zip_longest()`
To prevent truncation, we can use `zip_longest()` from the `itertools` library. It pads shorter sequences with a placeholder value like `"N/A"` so no data is lost.

```python
from itertools import zip_longest

students = ["Alice", "Bob", "Charlie", "Diana"]
scores = [85, 92]

# Fills missing values with 'N/A' instead of dropping them
full_pairs = list(zip_longest(students, scores, fillvalue="N/A"))
print("Padded Result:", full_pairs)
```
