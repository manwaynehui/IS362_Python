```python
def first_and_last_same(number_list):
    if number_list[0] == number_list[-1]:
        return True
    else:
        return False

numbers_x = [10, 20, 30, 40, 10]
numbers_y = [75, 65, 35, 75, 30]

print(f"Given list: {numbers_x} \nResult is {first_and_last_same(numbers_x)}")
print(f"Given list: {numbers_y} \nResult is {first_and_last_same(numbers_y)}")
```

```python
# pip install pandas
import pandas as pd

# Part a: Use a dictionary to create four fields:
Student_dictionary = {
    "Student": ["Tommy", "Fred", "Gail", "Man Wayne", "Randy", "Ariel"],
    "Hair Color": ["Brown", "Grey", "Blonde", "Black", "Black", "Brown"],
    "Eye Color": ["Hazel", "Green", "Blue", "Brown", "Hazel", "Blue"],
    "Weight in Kg": [60, 80, 76, 60, 88, 85],
}

# Part b: Index the dataframe
student_dataframe = pd.DataFrame(Student_dictionary)

# Part c: Print out the entire dataframe
print("Part C: Print out the entire dataframe \n")
print(student_dataframe)
print(f"{'==' * 30}\n")

# Part d: Print out hair color column as Pandas Series
print("Part D: Print out hair color column as Pandas Series\n")
print(student_dataframe["Hair Color"]) 
print(f"{'==' * 30}\n")


# Part e: Print out eye color column as Pandas DataFrame
print("Part E: Print out eye color column as Pandas DataFrame\n")
print(student_dataframe[["Eye Color"]]) 
print(f"{'==' * 30}\n")

# Part f: Print out DataFrame with student and weight columns
print("Part F: Print out DataFrame with student and weight columns\n")
print(student_dataframe[["Student", "Weight in Kg"]])
print(f"{'==' * 30}\n")

# Part g: Print out the second through fourth observations
print("Part G: Print out the second through fourth observations\n")
print(student_dataframe.iloc[1:4])
print(f"{'==' * 30}\n")
```
