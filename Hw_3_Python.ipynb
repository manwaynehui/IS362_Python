## Suppose a rider notes down at the end of each day the odometere log.
Day 1 is 55
Day 2 is 120
Fill out data to 10 days to see how our table functions.

```python
import pandas as pd

odometer_log = {
    "Day 1": 55,
    "Day 2": 120,
    "Day 3": 142,
    "Day 4": 173,
    "Day 5": 210,
    "Day 6": 275,
    "Day 7": 340,
    "Day 8": 432,
    "Day 9": 533,
    "Day 10": 601
}
```

## Convert odometer into a pandas series
Use `pd.series`

```python
cumulative_miles = pd.Series(odometer_log, name="Cumulative miles")
cumulative_miles.index.name = "Day"
```

## Calculate daily miles ridden
Calculates the  difference between consecutive in the Series Day_{n} - Day_{n-1}. This converts total cumulative mileage into daily miles ridden.

Because Day 1 has no previous day to subtract from, `diff()` produces `NaN` for the first entry which we dont want. 

How to fix the issue:
`.fillna()` -> Searches series for any 'NaN' and replaces with the value inside of parathesis.
`cumulative_miles.iloc[0]` -> Looks at first item in the Series which is `55`.

```python
daily_miles = cumulative_miles.diff().fillna(cumulative_miles)
daily_miles.name = "Daily miles"
```

## Combine into a Summary Table

We concatenate `cumulative_miles` and `daily_miles` along `axis=1` to form a final DataFrame that displays both the running odometer reading and the exact daily distance ridden side-by-side.

```python
# Combine into a summary DataFrame
ride_summary = pd.concat([cumulative_miles, daily_miles], axis=1)
ride_summary
```
