## Objetive

 Connect to the New York Times Article Search API, query live article data related to "trains" and convert the returned JSON reponse into a structure pandas DataFrame.


```python
import os
import requests
import pandas as pd
from dotenv import load_dotenv

load_dotenv()
API_KEY = os.environ.get("NYT_API_KEY")
```

`load_dotenv()` and `API_KEY = os.environ.get("NYT_API_KEY")`:
* Reads our hidden `.env` file and securely loads the NYT API key without exposing it in the notebook.
* Uses `.gitignore` to keep the `.env` file privates and prevents from uploading to github.

```python

url = "https://api.nytimes.com/svc/search/v2/articlesearch.json"
params = {
    "api-key": API_KEY,
    "q": "Trains"
}
```

`url`:
* Sets the NYT Article Search web address (endpoint) that we want to connect to.

`params`:
* Specifies the search details we are sending along with the request, including our API key and the search keyword `"Trains"`.

```python
response = requests.get(url, params=params)
response.raise_for_status()
data = response.json()
```

`response = requests.get(url, params=params)`:
* Sends the request to the NYT server and fetches the matching article data.

`data = response.json()`:
* Converts the raw JSON response from the server into a readable Python.

```python
articles = data['response']['docs']
df = pd.json_normalize(articles)
```

`articles = data['response']['docs']`:
* Opens the `'response'` section of the data and grabs just the list of articles (`'docs'`), ignoring the other extra data.

`df = pd.json_normalize(articles)`:
* Turns the raw article data into a clean pandas DataFrame with rows and columns.


```python
df[['abstract', 'pub_date', 'section_name', 'news_desk']].head()
```

`df[['abstract', 'pub_date', 'section_name', 'news_desk']].head()`:
* Selects only the 4 most useful columns (article summary, publish date, section, and news desk) instead of all 28 raw columns.
* Displays the first 5 rows as a clean, easy-to-read table.
